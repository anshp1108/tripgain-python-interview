```python
from __future__ import annotations
import os
import pandas as pd
from typing import Dict, Any
df = pd.read_csv("matches.csv")
```

```python
#Q1: 
print("\n\nQuestion 1: ")
total_matches = len(df)
columns = list(df.columns)
describe_all = df.describe(include="all")
print(f"Question 1 :")
print(f"\n\nTotal matches: {total_matches}")
print(f"\n\nColumns: {columns}")
print(f"\n\nFirst 5 rows: ")
print("\n")
print(df.head(5))
print(f"\n\nDescribe : {describe_all}")
```

```python
#Q2
print("\n\nQuestion 2: ")
final_ball = df[(df["win_by_runs"] == 1) | (df["win_by_wickets"] == 1)]
pom_counts = final_ball["player_of_match"].value_counts(dropna=True)
pom_top = pom_counts.head(1)
print(f"Final ball count :  {int(len(final_ball))}")
	
print(f"Top Player of the Match in final ball games: {pom_top.index[0]} ({pom_top.values[0]} times)")
```

```python
# Q3: Wankhede Stadium wins by runs vs wickets
mask_wankhede = df["venue"].str.contains("Wankhede", case=False, na=False)
wankhede = df[mask_wankhede]
wankhede_by_runs = int((wankhede["win_by_runs"] > 0).sum())
wankhede_by_wickets = int((wankhede["win_by_wickets"] > 0).sum())
print("\n\nQuestion 3: ")
print(f"wankhede_by_runs : {wankhede_by_runs}")
print(f"wankhede_by_wickets : {wankhede_by_wickets}")
if wankhede_by_runs>wankhede_by_wickets:
	print("It is more common to win by batting first")
else:
	print("It is more common to win by bowling first")
```

```python
# Q4: team with highest wins where win_by_runs > 50big_win = df[df["win_by_runs"] > 50]
big_win = df[df["win_by_runs"] > 50]
big_win_counts = big_win["winner"].value_counts()
big_win_top = big_win_counts.head(1)
print("\n\nQuestion 4: ")
print(f"Team with highest wins (win_by_runs > 50): {big_win_top.index[0]} ({big_win_top.values[0]} wins)")
```

```python
# Q5: toss winner chose to bat (set a target) and won the match
toss_set_target_and_won = df[(df["toss_winner"] == df["winner"]) & (df["toss_decision"].str.lower() == "bat")]
count_toss_set_target_and_won = len(toss_set_target_and_won)
print("\n\nQuestion 5: ")
print(f"Count of toss winner chose to bat and won: {count_toss_set_target_and_won}")
```

```python
# Q6: which umpire (umpire1 or umpire2) officiated more matches involving KKR
mask_kkr = (df["team1"] == "Kolkata Knight Riders") | (df["team2"] == "Kolkata Knight Riders")

kkr_matches = df[mask_kkr]
ump1_count = int(kkr_matches["umpire1"].dropna().shape[0])
ump2_count = int(kkr_matches["umpire2"].dropna().shape[0])
print("\n\nQuestion 6: ")
if ump1_count > ump2_count:
	print("umpire1 has more officiated matches involving KKR.")
else:
	print("umpire2 has officiated more matches involving the Kolkata Knight Riders.")
```
